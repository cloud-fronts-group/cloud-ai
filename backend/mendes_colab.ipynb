# Mendes AI — Colab backend

Runs **Ollama** and the **Mendes AI router** (`backend/server.py`) on this Colab GPU and exposes them through a Cloudflare quick tunnel.

1. *Runtime → Change runtime type → GPU* (L4 or A100 recommended for the default 27B/30B models).
2. *Runtime → Run all*.
3. The last cell prints `MENDES_URL` and `MENDES_TOKEN`. In VS Code open *Settings → Mendes AI* and paste them into **Tunnel Url** and **Token**.

Keep this notebook running while you work. The tunnel URL changes every time the runtime restarts.


```python
#@title 1 · Configuration
CODE_MODEL   = "qwen3:30b"   #@param {type:"string"}
VISION_MODEL = "gemma3:27b"  #@param {type:"string"}
REPO         = "https://github.com/cloud-fronts-group/mendes-ai"  #@param {type:"string"}

import os, secrets, subprocess, time, re, urllib.request
os.environ["MENDES_CODE_MODEL"] = CODE_MODEL
os.environ["MENDES_VISION_MODEL"] = VISION_MODEL
os.environ["MENDES_TOKEN"] = os.environ.get("MENDES_TOKEN") or secrets.token_urlsafe(24)
os.environ["OLLAMA_HOST"] = "http://127.0.0.1:11434"
!nvidia-smi --query-gpu=name,memory.total --format=csv
```

```python
#@title 2 · Install and start Ollama
!curl -fsSL https://ollama.com/install.sh | sh > /tmp/ollama-install.log 2>&1 && echo "ollama installed"
ollama_proc = subprocess.Popen(
    ["ollama", "serve"],
    stdout=open("/tmp/ollama.log", "w"), stderr=subprocess.STDOUT,
    env={**os.environ, "OLLAMA_HOST": "127.0.0.1:11434", "OLLAMA_KEEP_ALIVE": "-1"},
)
for _ in range(60):
    try:
        urllib.request.urlopen("http://127.0.0.1:11434/api/tags", timeout=2)
        print("ollama ready"); break
    except Exception:
        time.sleep(1)
else:
    print(open("/tmp/ollama.log").read())
```

```python
#@title 3 · Pull the models (several minutes the first time)
!ollama pull {CODE_MODEL}
!ollama pull {VISION_MODEL}
!ollama list
```

```python
#@title 4 · Fetch the router
!rm -rf /content/mendes-ai && git clone --depth 1 {REPO} /content/mendes-ai
!pip -q install -r /content/mendes-ai/backend/requirements.txt
```

```python
#@title 5 · Start the Mendes AI router
server_proc = subprocess.Popen(
    ["python", "/content/mendes-ai/backend/server.py"],
    stdout=open("/tmp/mendes.log", "w"), stderr=subprocess.STDOUT,
    env={**os.environ, "MENDES_HOST": "127.0.0.1", "MENDES_PORT": "8000"},
)
for _ in range(30):
    try:
        req = urllib.request.Request("http://127.0.0.1:8000/health",
                                     headers={"Authorization": "Bearer " + os.environ["MENDES_TOKEN"]})
        print(urllib.request.urlopen(req, timeout=2).read().decode()); break
    except Exception:
        time.sleep(1)
else:
    print(open("/tmp/mendes.log").read())
```

```python
#@title 6 · Open the Cloudflare tunnel and print your VS Code settings
!curl -fsSL -o /usr/local/bin/cloudflared https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64 && chmod +x /usr/local/bin/cloudflared
tunnel_proc = subprocess.Popen(
    ["cloudflared", "tunnel", "--url", "http://127.0.0.1:8000", "--no-autoupdate"],
    stdout=open("/tmp/cloudflared.log", "w"), stderr=subprocess.STDOUT,
)
url = None
for _ in range(60):
    m = re.search(r"https://[a-z0-9-]+\.trycloudflare\.com", open("/tmp/cloudflared.log").read())
    if m:
        url = m.group(0); break
    time.sleep(1)
os.environ["MENDES_URL"] = url or ""
print("=" * 64)
print("MENDES_URL=" + str(url))
print("MENDES_TOKEN=" + os.environ["MENDES_TOKEN"])
print("=" * 64)
print("VS Code -> Settings -> Mendes AI -> paste Tunnel Url and Token. Keep this notebook running.")
```

```python
#@title 7 · Verify through the tunnel
!curl -s -H "Authorization: Bearer $MENDES_TOKEN" "$MENDES_URL/health" | python -m json.tool
```
